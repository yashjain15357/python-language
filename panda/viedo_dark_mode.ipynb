```python
import cv2
import numpy as np

# Load video
cap = cv2.VideoCapture(r"C:\Users\YASH JAIN\Documents\New Folder\OneDrive\Videos\Captures\Flask Tutorial for Web Development and Data Science _ Sagar Chouksey - YouTube - Google Chrome 2025-08-12 10-38-13.mp4")
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
out = cv2.VideoWriter('dark_theme_video.mp4', fourcc, fps, (width, height))

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    # Convert to HSV to work with brightness
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    h, s, v = cv2.split(hsv)
    
    # Create mask for "light" areas (bright backgrounds)
    light_mask = cv2.inRange(v, 200, 255)  # threshold for bright areas
    
    # Invert colors in those bright areas
    inverted = cv2.bitwise_not(frame)
    
    # Combine inverted bright areas with original dark areas
    dark_theme_frame = frame.copy()
    dark_theme_frame[light_mask > 0] = inverted[light_mask > 0]
    
    out.write(dark_theme_frame)

cap.release()
out.release()
print("✅ Dark theme video saved as 'dark_theme_video.mp4'")
```

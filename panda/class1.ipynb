<h1>Create data frame<h1>

```python
import pandas as pd
import numpy as np
```

<h2>Dict of list<h2>


```python
dict = {
    "Name":["yash", "jobin","neeaj"],
    "Age":[19,20,21],
    "Class":["5th sem","5th sem","5th sem"]
}
table = pd.DataFrame(dict)
print(table)
```

<h2>list of dict<h2>


```python
arr=[{"Name":"yash",
     "Age":19,
     "Class":"5th sem"},

     {"Name":"jobin",
     "Age":19,
     "Class":"5th sem"},

     {"Name":"neeraj",
     "Age":19,
     "Class":"5th sem"}
     
     ]
table = pd.DataFrame(arr)
print(table)
```

<h2>list of list<h2>


```python
arr = [["yash", "jobin","neeaj"],
    [19,20,21],
    ["5th sem","5th sem","5th sem"]]
table = pd.DataFrame(arr,index=["Name","Age","Class"])
print(table)
print(type(table))
```

<h2>numpy array<h2>


```python
df =np.array([["yash", "jobin","neeaj"],
    [19,20,21],
    ["5th sem","5th sem","5th sem"]]) 
table = pd.DataFrame(df,index=["Name","Age","Class"])
print(table)
```
